```python
df["Order Date"] = pd.to_datetime(df["Order Date"], dayfirst=True)
df["Ship Date"] = pd.to_datetime(df["Ship Date"], dayfirst=True)

df["Lead Time"] = (df["Ship Date"] - df["Order Date"]).dt.days

df[["Order Date", "Ship Date", "Lead Time"]].head()
```

```python
df["Lead Time"].describe()
df[df["Lead Time"] < 0]
```

```python
df = df[df["Lead Time"] >= 0].copy()
```

```python
factory_map = {
    "Wonka Bar - Nutty Crunch Surprise": "Lot's O' Nuts",
    "Wonka Bar - Fudge Mallows": "Lot's O' Nuts",
    "Wonka Bar -Scrumdiddlyumptious": "Lot's O' Nuts",
    "Wonka Bar - Milk Chocolate": "Wicked Choccy's",
    "Wonka Bar - Triple Dazzle Caramel": "Wicked Choccy's",
    "Laffy Taffy": "Sugar Shack",
    "SweeTARTS": "Sugar Shack",
    "Nerds": "Sugar Shack",
    "Fun Dip": "Sugar Shack",
    "Fizzy Lifting Drinks": "Sugar Shack",
    "Everlasting Gobstopper": "Secret Factory",
    "Hair Toffee": "The Other Factory",
    "Lickable Wallpaper": "Secret Factory",
    "Wonka Gum": "Secret Factory",
    "Kazookles": "The Other Factory"
}

df["Factory"] = df["Product Name"].map(factory_map)
```

```python
print(df['Factory'].isna().sum())

print(df['Factory'].value_counts())
```

```python
df['Route'] = df['Factory'] + '->' + df['State/Province']

df[["Factory", "State/Province", "Route"]].head()
```

```python
#Route-wise KPI

route_kpi = df.groupby('Route').agg(
    Shipments=('Order ID', 'count'),
    Avg_Lead_Time=('Lead Time', 'mean'),
    Lead_Time_Std=('Lead Time', 'std'),
).reset_index()

route_kpi.head()
```

```python
# Delay threshold
Delayed_threshold = 5

df['Delayed'] = df['Lead Time'] > Delayed_threshold
```

```python
# Route delay frequency

delay_rate = df.groupby('Route')['Delayed'].mean().mul(100).reset_index(name='Delayed_Frequency')
route_kpi = route_kpi.merge(delay_rate, on='Route')

route_kpi.head()
```

```python
# Efficiency Score

route_kpi['Efficiency_Score'] = (
    100*(route_kpi['Avg_Lead_Time'].max() - route_kpi['Avg_Lead_Time']) / 
    (route_kpi['Avg_Lead_Time'].max() - route_kpi['Avg_Lead_Time'].min())
).round(2)

route_kpi.head()
```

```python
top_routes = route_kpi.nsmallest(10, "Avg_Lead_Time")
bottom_routes = route_kpi.nlargest(10, "Avg_Lead_Time")

print(top_routes)
print(bottom_routes)
```

```python
# Ship Mode performance

ship_mode_kpi = df.groupby("Ship Mode").agg(
    Shipments = ('Order ID', 'count'),
    Avg_Lead_Time = ('Lead Time', 'mean'),
).reset_index()

ship_mode_kpi.head()
```

```python
# Region performance
region_kpi = df.groupby("Region").agg(
    Shipments=("Order ID", "count"),
    Avg_Lead_Time=("Lead Time", "mean")
).reset_index()

region_kpi
```

```python
# State performance
state_kpi = df.groupby("State/Province").agg(
    Shipments=("Order ID", "count"),
    Avg_Lead_Time=("Lead Time", "mean")
).reset_index()

state_kpi = state_kpi.sort_values("Avg_Lead_Time", ascending=False)
state_kpi.head(10)
```

```python
# High-volume + slow states
volume_limit = state_kpi["Shipments"].median()

bottlenecks = state_kpi[
    (state_kpi["Shipments"] >= volume_limit)
].sort_values("Avg_Lead_Time", ascending=False)

bottlenecks.head(10)
```

```python
#Route coordinates

route_map = df[df["State/Province"].isin(state_coords)].copy()
route_map['Customer_Lat'] = route_map['State/Province'].map(lambda x: state_coords[x][0])
route_map['Customer_Lon'] = route_map['State/Province'].map(lambda x: state_coords[x][1])
route_map['Factory_Lat'] = route_map['Factory'].map(lambda x: factory_coords[x][0])
route_map['Factory_Lon'] = route_map['Factory'].map(lambda x: factory_coords[x][1])
```

```python
# Unique Route sample

routes = route_map[["Factory", "Factory_Lat", "Factory_Lon",
                    "State/Province", "Customer_Lat", "Customer_Lon"]].drop_duplicates()
```

```python
fig = px.scatter_map(
    routes,
    lat="Customer_Lat", lon="Customer_Lon",
    hover_name="State/Province",
    hover_data=["Factory"],
    color="Factory",
    zoom=3,
    height=600,
    title="Factory-to-Customer Shipping Routes"
)
fig.show()
```

```python
#Route lines map

import plotly.graph_objects as go

fig = go.Figure()

for _, r in routes.iterrows():
    fig.add_trace(go.Scattergeo(
        lon=[r["Factory_Lon"], r["Customer_Lon"]],
        lat=[r["Factory_Lat"], r["Customer_Lat"]],
        mode="lines",
        line=dict(width=1),
        hoverinfo="text",
        text=f'{r["Factory"]} → {r["State/Province"]}'
    ))

fig.update_layout(
    title="Factory-to-Customer Shipping Routes",
    geo=dict(scope="usa", showland=True),
    height=600
)

fig.show()
```

```python
# Factory locations highlight

factory_df = pd.DataFrame([
    {"Factory": k, "Lat": v[0], "Lon": v[1]}
    for k, v in factory_coords.items()
])

fig.add_trace(go.Scattergeo(
    lon=factory_df["Lon"],
    lat=factory_df["Lat"],
    mode="markers+text",
    text=factory_df["Factory"],
    textposition="top center",
    marker=dict(size=10),
    name="Factories"
))

fig.show()
```
